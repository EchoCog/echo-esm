# Bio-Cognitive Accounting with ESM3

This notebook demonstrates the revolutionary **Bio-Cognitive Accounting Framework** that transforms protein analysis by mapping traditional accounting concepts to biological systems.

## Core Concept Mapping

The framework creates an intuitive bridge between accounting and biology:

| Accounting Concept | Bio-Cognitive Translation | Description |
|-------------------|---------------------------|-------------|
| **Accounts** | **Proteins** | Each protein becomes a cognitive account with unique properties |
| **Transactions** | **Amino Acids** | Each amino acid is a cognitive transaction with metadata |
| **Balances** | **Structures** | Protein structures represent account balance states |
| **Ledgers** | **Sequences** | Amino acid sequences form transaction ledgers |
| **Performance** | **Functions** | Protein functions become performance metrics |
| **Trial Balance** | **System Validation** | PLN-based proof of protein set consistency |

## Cognitive Account Types

The framework supports multiple cognitive account types with increasing intelligence:

- **Traditional (Flag: 1)**: Basic protein representation with AtomSpace storage
- **Adaptive (Flag: 2)**: Learning-enabled proteins that improve over time
- **Predictive (Flag: 4)**: Forward-looking analysis with uncertainty quantification
- **Multimodal (Flag: 8)**: Multi-input processing (sequence, structure, function)
- **Attention-Driven (Flag: 16)**: Economic attention allocation for resource optimization

Account types can be combined using bitwise OR operations for maximum flexibility.

## Demo Implementation

Since we may encounter import issues in Colab, let's create a demonstration implementation that shows the core concepts:

```python
# Demo Bio-Cognitive Framework Implementation
import uuid
from enum import Flag
from dataclasses import dataclass
from typing import Dict, List, Optional, Any

class DemoCognitiveAccountType(Flag):
    """Demo cognitive account type flags"""
    TRADITIONAL = 1
    ADAPTIVE = 2
    PREDICTIVE = 4
    MULTIMODAL = 8
    ATTENTION_DRIVEN = 16

@dataclass
class DemoProtein:
    """Demo protein representation"""
    sequence: str
    name: str = None
    
    def __post_init__(self):
        if self.name is None:
            self.name = f"Protein_{str(uuid.uuid4())[:8]}"

@dataclass 
class DemoTruthValue:
    """Demo truth value for cognitive reasoning"""
    strength: float
    confidence: float
    
    def __post_init__(self):
        self.strength = max(0.0, min(1.0, self.strength))
        self.confidence = max(0.0, min(1.0, self.confidence))

@dataclass
class DemoAttentionValue:
    """Demo attention value for ECAN"""
    sti: float  # Short-term importance
    lti: float  # Long-term importance 
    vlti: float # Very long-term importance

class DemoCognitiveProtein:
    """Demo cognitive protein with accounting capabilities"""
    
    def __init__(self, protein: DemoProtein, cognitive_type: DemoCognitiveAccountType = DemoCognitiveAccountType.TRADITIONAL):
        self.protein = protein
        self.cognitive_type = cognitive_type
        self.account_id = f"ACC_{protein.name}"
        
        # Accounting components
        self.transactions = []  # Amino acid transactions
        self.balance_state = self._calculate_initial_balance()
        self.performance_metrics = {}
        self.attention_value = DemoAttentionValue(100.0, 50.0, 0.0)
        
        # Learning components (if adaptive)
        self.learned_patterns = {} if DemoCognitiveAccountType.ADAPTIVE in cognitive_type else None
        self.prediction_models = {} if DemoCognitiveAccountType.PREDICTIVE in cognitive_type else None
        
        # Process amino acid transactions
        self._process_amino_acid_transactions()
    
    def _calculate_initial_balance(self) -> Dict[str, float]:
        """Calculate initial balance state from protein structure"""
        sequence = self.protein.sequence
        
        # Simplified balance calculation based on amino acid properties
        hydrophobic_count = sum(1 for aa in sequence if aa in 'AILMFPWV')
        hydrophilic_count = sum(1 for aa in sequence if aa in 'RNDQEHKSYT')
        charged_count = sum(1 for aa in sequence if aa in 'RKDENH')
        
        total_length = len(sequence)
        
        return {
            'hydrophobic_balance': hydrophobic_count / total_length if total_length > 0 else 0,
            'hydrophilic_balance': hydrophilic_count / total_length if total_length > 0 else 0,
            'charge_balance': charged_count / total_length if total_length > 0 else 0,
            'length_score': min(1.0, total_length / 500.0),  # Normalized to typical protein length
            'overall_balance': np.mean([hydrophobic_count, hydrophilic_count, charged_count]) / total_length if total_length > 0 else 0
        }
    
    def _process_amino_acid_transactions(self):
        """Process each amino acid as a cognitive transaction"""
        aa_properties = {
            'A': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': 1.8},
            'R': {'type': 'basic', 'charge': 1, 'hydrophobicity': -4.5},
            'N': {'type': 'polar', 'charge': 0, 'hydrophobicity': -3.5},
            'D': {'type': 'acidic', 'charge': -1, 'hydrophobicity': -3.5},
            'C': {'type': 'sulfur', 'charge': 0, 'hydrophobicity': 2.5},
            'Q': {'type': 'polar', 'charge': 0, 'hydrophobicity': -3.5},
            'E': {'type': 'acidic', 'charge': -1, 'hydrophobicity': -3.5},
            'G': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': -0.4},
            'H': {'type': 'basic', 'charge': 0.5, 'hydrophobicity': -3.2},
            'I': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': 4.5},
            'L': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': 3.8},
            'K': {'type': 'basic', 'charge': 1, 'hydrophobicity': -3.9},
            'M': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': 1.9},
            'F': {'type': 'aromatic', 'charge': 0, 'hydrophobicity': 2.8},
            'P': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': -1.6},
            'S': {'type': 'polar', 'charge': 0, 'hydrophobicity': -0.8},
            'T': {'type': 'polar', 'charge': 0, 'hydrophobicity': -0.7},
            'W': {'type': 'aromatic', 'charge': 0, 'hydrophobicity': -0.9},
            'Y': {'type': 'aromatic', 'charge': 0, 'hydrophobicity': -1.3},
            'V': {'type': 'nonpolar', 'charge': 0, 'hydrophobicity': 4.2}
        }
        
        for position, amino_acid in enumerate(self.protein.sequence):
            if amino_acid in aa_properties:
                transaction = {
                    'position': position,
                    'amino_acid': amino_acid,
                    'properties': aa_properties[amino_acid],
                    'transaction_id': f"TXN_{self.account_id}_{position}_{amino_acid}",
                    'confidence': DemoTruthValue(0.9, 0.8),  # High confidence in amino acid identity
                    'attention_weight': 1.0 / len(self.protein.sequence)  # Equal initial attention
                }
                self.transactions.append(transaction)
    
    def get_account_summary(self) -> Dict[str, Any]:
        """Get comprehensive account summary"""
        return {
            'account_id': self.account_id,
            'protein_name': self.protein.name,
            'cognitive_type': {
                'value': self.cognitive_type.value,
                'flags': [flag.name for flag in DemoCognitiveAccountType if flag in self.cognitive_type]
            },
            'sequence_length': len(self.protein.sequence),
            'transaction_count': len(self.transactions),
            'balance_state': self.balance_state,
            'attention_value': {
                'sti': self.attention_value.sti,
                'lti': self.attention_value.lti,
                'vlti': self.attention_value.vlti
            },
            'capabilities': {
                'learning_enabled': self.learned_patterns is not None,
                'prediction_enabled': self.prediction_models is not None,
                'multimodal_processing': DemoCognitiveAccountType.MULTIMODAL in self.cognitive_type,
                'attention_driven': DemoCognitiveAccountType.ATTENTION_DRIVEN in self.cognitive_type
            }
        }
    
    def predict_stability(self) -> Dict[str, float]:
        """Predict protein stability using cognitive model"""
        if DemoCognitiveAccountType.PREDICTIVE not in self.cognitive_type:
            return {'error': 'Predictive capabilities not enabled'}
        
        # Simplified stability prediction based on balance state
        balance = self.balance_state
        
        # Stability factors
        hydrophobic_stability = balance['hydrophobic_balance'] * 0.4
        charge_stability = (1.0 - abs(balance['charge_balance'] - 0.2)) * 0.3  # Optimal charge balance around 20%
        length_stability = balance['length_score'] * 0.3
        
        predicted_stability = hydrophobic_stability + charge_stability + length_stability
        
        # Add some uncertainty
        uncertainty = 0.1 + (0.1 * np.random.random())
        
        return {
            'predicted_stability': predicted_stability,
            'uncertainty': uncertainty,
            'confidence_interval': [
                max(0.0, predicted_stability - uncertainty),
                min(1.0, predicted_stability + uncertainty)
            ],
            'factors': {
                'hydrophobic_contribution': hydrophobic_stability,
                'charge_contribution': charge_stability,
                'length_contribution': length_stability
            }
        }

print("✓ Demo Bio-Cognitive Framework implementation ready")
```

## Example 3: Multi-Protein Trial Balance

Let's create multiple proteins and demonstrate the trial balance concept - ensuring the protein set maintains consistency:

```python
# Create a portfolio of proteins for trial balance demonstration
protein_portfolio = {
    'enzyme': DemoProtein(
        sequence="MSPVLVQMSPKGSPQAAALLLWLWLWLWGPGPGPGAPDAPDAPDAPVPVPVPVPGQGQGQGQHQHQHQHQ",
        name="DemoEnzyme"
    ),
    'structural': DemoProtein(
        sequence="MGAAALLLWLWLWLWGGGGGGCCCCCCDDDDDDRRRRRKKKKKKEEEEEEHHHHHHSSSSSSTTTTTT",
        name="StructuralProtein"
    ),
    'regulatory': DemoProtein(
        sequence="MKKKKKRRRRRHHHHHEEEEEEDDDDDLLLLLIIIIIVVVVVFFFFWWWWWYYYYYY",
        name="RegulatoryProtein"
    )
}

# Create cognitive accounts for each protein
cognitive_portfolio = {}
for protein_type, protein in protein_portfolio.items():
    # Use different cognitive types for different protein types
    if protein_type == 'enzyme':
        cognitive_type = DemoCognitiveAccountType.ADAPTIVE | DemoCognitiveAccountType.PREDICTIVE
    elif protein_type == 'structural':
        cognitive_type = DemoCognitiveAccountType.TRADITIONAL | DemoCognitiveAccountType.ATTENTION_DRIVEN
    else:  # regulatory
        cognitive_type = DemoCognitiveAccountType.MULTIMODAL | DemoCognitiveAccountType.PREDICTIVE
    
    cognitive_portfolio[protein_type] = DemoCognitiveProtein(protein, cognitive_type)

# Create trial balance summary
print("📊 Protein Portfolio Trial Balance")
print("=" * 50)

# Portfolio summary table
portfolio_data = []
total_transactions = 0
total_attention_sti = 0
total_balance_score = 0

for protein_type, cognitive_protein in cognitive_portfolio.items():
    summary = cognitive_protein.get_account_summary()
    
    portfolio_data.append({
        'Protein Type': protein_type.capitalize(),
        'Account ID': summary['account_id'],
        'Cognitive Flags': len(summary['cognitive_type']['flags']),
        'Sequence Length': summary['sequence_length'],
        'Transactions': summary['transaction_count'],
        'Overall Balance': summary['balance_state']['overall_balance'],
        'STI Attention': summary['attention_value']['sti']
    })
    
    total_transactions += summary['transaction_count']
    total_attention_sti += summary['attention_value']['sti']
    total_balance_score += summary['balance_state']['overall_balance']

# Display portfolio as DataFrame
portfolio_df = pd.DataFrame(portfolio_data)
print(portfolio_df.to_string(index=False))

print("\n📈 Portfolio Totals:")
print(f"Total Proteins: {len(cognitive_portfolio)}")
print(f"Total Transactions: {total_transactions}")
print(f"Total STI Attention: {total_attention_sti:.1f}")
print(f"Average Balance Score: {total_balance_score / len(cognitive_portfolio):.3f}")

# Trial balance validation (simplified)
balance_variance = np.var([p['Overall Balance'] for p in portfolio_data])
attention_variance = np.var([p['STI Attention'] for p in portfolio_data])

print("\n✅ Trial Balance Validation:")
print(f"Balance Variance: {balance_variance:.6f} (lower is better)")
print(f"Attention Variance: {attention_variance:.2f}")
print(f"Portfolio Consistency: {'GOOD' if balance_variance < 0.01 else 'NEEDS REVIEW'}")
```

## Example 4: Attention-Driven Economic Analysis

Demonstrate the Economic Attention Allocation (ECAN) system that manages computational resources like an economy:

```python
# Simulate ECAN attention economy over time
class AttentionEconomy:
    def __init__(self, initial_funds=1000.0):
        self.sti_funds = initial_funds
        self.lti_funds = initial_funds * 0.5
        self.wage_rate = 2.0
        self.rent_rate = 0.02
        self.cycle_count = 0
        
    def allocate_attention_cycle(self, proteins):
        """Simulate one attention allocation cycle"""
        self.cycle_count += 1
        
        # Pay wages based on activity
        for protein in proteins:
            activity_level = np.random.random()  # Simulate activity
            wage = activity_level * self.wage_rate
            
            if self.sti_funds >= wage:
                protein.attention_value.sti += wage
                self.sti_funds -= wage
        
        # Collect rent (attention decay)
        for protein in proteins:
            rent = protein.attention_value.sti * self.rent_rate
            protein.attention_value.sti = max(0, protein.attention_value.sti - rent)
            self.sti_funds += rent * 0.5  # Return some to economy
        
        # LTI investment (long-term importance)
        for protein in proteins:
            if protein.attention_value.sti > 150:  # High STI proteins get LTI
                lti_investment = min(10.0, self.lti_funds * 0.1)
                protein.attention_value.lti += lti_investment
                self.lti_funds -= lti_investment

# Run attention economy simulation
economy = AttentionEconomy()
proteins_list = list(cognitive_portfolio.values())

# Track attention over time
attention_history = {protein_type: {'sti': [], 'lti': [], 'cycles': []} 
                    for protein_type in cognitive_portfolio.keys()}
fund_history = {'sti_funds': [], 'lti_funds': [], 'cycles': []}

# Run simulation for 50 cycles
for cycle in range(50):
    economy.allocate_attention_cycle(proteins_list)
    
    # Record attention values
    for protein_type, cognitive_protein in cognitive_portfolio.items():
        attention_history[protein_type]['sti'].append(cognitive_protein.attention_value.sti)
        attention_history[protein_type]['lti'].append(cognitive_protein.attention_value.lti)
        attention_history[protein_type]['cycles'].append(cycle)
    
    # Record fund levels
    fund_history['sti_funds'].append(economy.sti_funds)
    fund_history['lti_funds'].append(economy.lti_funds)
    fund_history['cycles'].append(cycle)

# Visualize attention economy dynamics
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))

# Plot 1: STI Attention over time
for protein_type in cognitive_portfolio.keys():
    ax1.plot(attention_history[protein_type]['cycles'], 
             attention_history[protein_type]['sti'], 
             label=protein_type.capitalize(), linewidth=2)
ax1.set_title('Short-Term Importance (STI) Over Time')
ax1.set_xlabel('Cycle')
ax1.set_ylabel('STI Value')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot 2: LTI Attention over time
for protein_type in cognitive_portfolio.keys():
    ax2.plot(attention_history[protein_type]['cycles'], 
             attention_history[protein_type]['lti'], 
             label=protein_type.capitalize(), linewidth=2)
ax2.set_title('Long-Term Importance (LTI) Over Time')
ax2.set_xlabel('Cycle')
ax2.set_ylabel('LTI Value')
ax2.legend()
ax2.grid(True, alpha=0.3)

# Plot 3: Economic Funds
ax3.plot(fund_history['cycles'], fund_history['sti_funds'], 
         label='STI Funds', linewidth=2, color='blue')
ax3.plot(fund_history['cycles'], fund_history['lti_funds'], 
         label='LTI Funds', linewidth=2, color='red')
ax3.set_title('Attention Economy Fund Levels')
ax3.set_xlabel('Cycle')
ax3.set_ylabel('Fund Level')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Plot 4: Final attention distribution
final_sti = [cognitive_portfolio[p].attention_value.sti for p in cognitive_portfolio.keys()]
final_lti = [cognitive_portfolio[p].attention_value.lti for p in cognitive_portfolio.keys()]
protein_names = [p.capitalize() for p in cognitive_portfolio.keys()]

x = np.arange(len(protein_names))
width = 0.35

ax4.bar(x - width/2, final_sti, width, label='STI', alpha=0.8)
ax4.bar(x + width/2, final_lti, width, label='LTI', alpha=0.8)
ax4.set_title('Final Attention Distribution')
ax4.set_xlabel('Protein')
ax4.set_ylabel('Attention Value')
ax4.set_xticks(x)
ax4.set_xticklabels(protein_names)
ax4.legend()
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.suptitle('Bio-Cognitive Attention Economy Dynamics', fontsize=16, y=1.02)
plt.show()

# Print final economy status
print("💰 Final Economy Status:")
print(f"STI Funds Remaining: {economy.sti_funds:.2f}")
print(f"LTI Funds Remaining: {economy.lti_funds:.2f}")
print(f"Total Cycles: {economy.cycle_count}")
print("\n🏆 Attention Winners:")
attention_ranking = [(p, cognitive_portfolio[p].attention_value.sti + cognitive_portfolio[p].attention_value.lti) 
                    for p in cognitive_portfolio.keys()]
attention_ranking.sort(key=lambda x: x[1], reverse=True)
for i, (protein_type, total_attention) in enumerate(attention_ranking, 1):
    print(f"{i}. {protein_type.capitalize()}: {total_attention:.2f} total attention")
```

## Example 5: Protein Structure Visualization with Cognitive Overlay

Let's visualize protein structures with cognitive accounting information overlaid:

```python
# Generate a simplified 3D protein structure visualization with cognitive data
def generate_protein_coords(sequence, protein_type='enzyme'):
    """Generate simplified 3D coordinates for demonstration"""
    n = len(sequence)
    
    # Different folding patterns for different protein types
    if protein_type == 'enzyme':
        # More compact, globular structure
        theta = np.linspace(0, 4*np.pi, n)
        radius = 10 + 5 * np.sin(theta/2)
        x = radius * np.cos(theta)
        y = radius * np.sin(theta)
        z = 2 * theta
    elif protein_type == 'structural':
        # More extended, fibrous structure
        t = np.linspace(0, 20, n)
        x = t
        y = 5 * np.sin(t)
        z = 5 * np.cos(t/2)
    else:  # regulatory
        # Intermediate structure
        t = np.linspace(0, 6*np.pi, n)
        x = 8 * np.cos(t)
        y = 8 * np.sin(t)
        z = t
    
    return np.column_stack([x, y, z])

# Create 3D visualization
fig = go.Figure()

colors = ['red', 'blue', 'green']
protein_types = list(cognitive_portfolio.keys())

for i, (protein_type, cognitive_protein) in enumerate(cognitive_portfolio.items()):
    coords = generate_protein_coords(cognitive_protein.protein.sequence, protein_type)
    
    # Color points based on attention values
    attention_normalized = cognitive_protein.attention_value.sti / 200.0  # Normalize to 0-1
    
    # Create trace for protein backbone
    fig.add_trace(go.Scatter3d(
        x=coords[:, 0],
        y=coords[:, 1],
        z=coords[:, 2],
        mode='lines+markers',
        line=dict(color=colors[i], width=4),
        marker=dict(
            size=3 + attention_normalized * 5,  # Size based on attention
            color=colors[i],
            opacity=0.6 + attention_normalized * 0.4
        ),
        name=f'{protein_type.capitalize()} (STI: {cognitive_protein.attention_value.sti:.1f})',
        text=[f'AA: {aa}<br>Pos: {j}<br>Attention: {attention_normalized:.3f}' 
              for j, aa in enumerate(cognitive_protein.protein.sequence)],
        hovertemplate='%{text}<extra></extra>'
    ))

# Update layout for 3D visualization
fig.update_layout(
    title='Bio-Cognitive Protein Structures<br><sub>Sphere size and opacity indicate attention values</sub>',
    scene=dict(
        xaxis_title='X Coordinate',
        yaxis_title='Y Coordinate', 
        zaxis_title='Z Coordinate',
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.5)
        )
    ),
    width=800,
    height=600
)

fig.show()

# Create cognitive accounting summary table
summary_data = []
for protein_type, cognitive_protein in cognitive_portfolio.items():
    summary = cognitive_protein.get_account_summary()
    
    # Calculate cognitive complexity score
    complexity_score = (
        len(summary['cognitive_type']['flags']) * 0.3 +
        summary['balance_state']['overall_balance'] * 0.4 +
        (summary['attention_value']['sti'] / 200.0) * 0.3
    )
    
    summary_data.append({
        'Account ID': summary['account_id'],
        'Protein Type': protein_type.capitalize(),
        'Cognitive Flags': ', '.join(summary['cognitive_type']['flags']),
        'Sequence Length': summary['sequence_length'],
        'Balance Score': f"{summary['balance_state']['overall_balance']:.3f}",
        'STI Attention': f"{summary['attention_value']['sti']:.1f}",
        'LTI Attention': f"{summary['attention_value']['lti']:.1f}",
        'Complexity Score': f"{complexity_score:.3f}",
        'Learning Enabled': '✅' if summary['capabilities']['learning_enabled'] else '❌',
        'Prediction Enabled': '✅' if summary['capabilities']['prediction_enabled'] else '❌'
    })

summary_df = pd.DataFrame(summary_data)
print("\n📋 Final Bio-Cognitive Accounting Summary")
print("=" * 80)
print(summary_df.to_string(index=False))
```
